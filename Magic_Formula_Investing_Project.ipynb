# Magic Formula Investing Project

Screens US-listed stocks with Joel Greenblatt's Magic Formula (earnings yield + return on capital) plus eight valuation ranks and two momentum ranks, and writes a ranked snapshot to `Results/`.

The logic lives in the `magic_formula/` package (so it can be tested). This notebook is a walkthrough on a small sample. The full monthly run is a command:

```bash
export SEC_USER_AGENT="Your Name your@email.com"   # the SEC asks scrapers to identify themselves
python -m magic_formula run                          # ~10K tickers, takes a couple of hours
```

*For education and research only; not investment advice.*

## 1. Pick tickers

Edit the list, or load the full SEC universe with `tickers.fetch_sec_tickers()` (needs `SEC_USER_AGENT`).

````python
from magic_formula import tickers

universe = tickers.normalize(["MSFT", "AAPL", "NVDA", "JNJ", "HPQ", "CROX", "EPAM", "GDDY", "YELP", "OMC"])
# universe = tickers.normalize(tickers.fetch_sec_tickers())
````

## 2. Scrape fundamentals and momentum (yfinance)

Drops companies under $200M market cap and the Utilities and Financial Services sectors.

````python
from magic_formula.scrape import scrape

raw, stats = scrape(universe)
print(stats.summary())
raw
````

## 3. Rank

`Final_Rank` orders stocks by a weighted sum of ranks (lower is better): Magic Formula x2, eight valuation ranks x1, two momentum ranks x0.5. See `magic_formula/ranking.py`.

````python
from magic_formula.ranking import rank_stocks

ranked = rank_stocks(raw)
ranked[["Company Name", "Ticker", "Market Cap", "MagicFormula_Rank", "Combined_Rank", "Final_Rank"]]
````
